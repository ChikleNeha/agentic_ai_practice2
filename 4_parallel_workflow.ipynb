```python
from langgraph.graph import StateGraph, START , END     
from typing import TypedDict
from langchain_ollama import ChatOllama
```

```python
class LLMState(TypedDict):
    sentence: str
    frnd_resp: str
    parent_resp: str
    therapists_resp: str
```

```python
model = ChatOllama(model="llama3.2:latest")
```

```python
def frnd(state: LLMState):
    stc = state['sentence']

    response = model.invoke(f"respond as a friend to the following sentence - keep it very short : {stc}").content

    return {"frnd_resp" : response}
```

```python
def parent(state: LLMState):
    stc = state['sentence']

    response = model.invoke(f"respond as a parent to the following sentence - keep it very short : {stc}").content

    return {"parent_resp" : response}
```

```python
def therapist(state: LLMState):
    stc = state['sentence']

    response = model.invoke(f"respond as a therapist to the following sentence - keep it very short : {stc}").content

    return {"therapist_resp" : response}
```

```python
def random(state: LLMState):
    pass
```

```python
graph = StateGraph(LLMState)

# adding nodes
graph.add_node("frnd",frnd)
graph.add_node("parent",parent)
graph.add_node("therapist",therapist)
graph.add_node("random",random)

# adding edges
graph.add_edge(START,"frnd")
graph.add_edge(START,"parent")
graph.add_edge(START,"therapist")

graph.add_edge("frnd", "random")
graph.add_edge("parent", "random")
graph.add_edge("therapist", "random")

graph.add_edge("random", END)

workflow = graph.compile()
```

```python
initial_state = {"sentence" : "I am feeling tired and overstimulated"}

final_state = workflow.invoke(initial_state)
```
